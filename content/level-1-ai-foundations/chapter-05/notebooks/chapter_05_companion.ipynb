## 1. Import Modul Logika dan Representasi
Memuat struktur Fact, Rule, KnowledgeBase, dan TruthTable.

```python
import sys, os
sys.path.append(os.path.abspath("../code"))
from facts_and_rules import KnowledgeBase, create_network_troubleshooting_kb
from truth_table import PropositionalLogic, generate_truth_table, check_entailment
from forward_chaining import ForwardChainingEngine
from backward_chaining import BackwardChainingEngine

kb = create_network_troubleshooting_kb()
print(f"KB Dimuat: {len(kb.facts)} Fakta Awal dan {len(kb.rules)} Aturan Logika.")
```

## 2. Pembangkit Tabel Kebenaran & Model Checking Entailment
Menguji apakah pengetahuan premis P dan kaidah P -> Q secara semantik menjamin kesimpulan Q (Modus Ponens).

```python
# Evaluasi Entailment: KB |= Q
vars_pq = ["P", "Q"]
kb_fn = lambda m: m["P"] and PropositionalLogic.IMPLIES(m["P"], m["Q"])
query_fn = lambda m: m["Q"]

entails = check_entailment(vars_pq, kb_fn, query_fn)
print("Hasil Uji Model Checking Entailment (KB |= Q):", entails)
```

## 3. Forward Chaining (Data-Driven Inference)
Mengeksekusi inferensi maju dari fakta-fakta gejala jaringan menuju kesimpulan status internet.

```python
fw_engine = ForwardChainingEngine(kb)
is_proven, new_facts = fw_engine.run(query="internet_inaccessible")
print("Query 'internet_inaccessible' Terbukti?:", is_proven)
print("Fakta Baru yang Ditarik:", new_facts)
```

## 4. Backward Chaining (Goal-Driven Proof)
Membuktikan hipotesis secara rekursif dari target tujuan ke fakta dasar.

```python
bw_engine = BackwardChainingEngine(kb)
proven = bw_engine.prove("internet_inaccessible")
print("Hipotesis 'internet_inaccessible' Terbukti?:", proven)
```
