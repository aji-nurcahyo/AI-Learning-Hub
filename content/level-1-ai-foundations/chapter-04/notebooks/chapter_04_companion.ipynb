## 1. Import Modul dan Benchmark Problem
Memuat modul fondasi pencarian dan graf klasik Romania.

```python
import sys, os
sys.path.append(os.path.abspath("../code"))
from search_framework import RomaniaMapProblem, GridNavigationProblem
from bfs_search import breadth_first_search
from dfs_search import depth_first_search
from uniform_cost_search import uniform_cost_search
from greedy_best_first_search import greedy_best_first_search
from astar_search import astar_search

romania = RomaniaMapProblem("Arad", "Bucharest")
print("Kota awal:", romania.initial_state, "-> Kota tujuan:", romania.goal_state)
print("Aksi dari Arad:", romania.actions("Arad"))
```

## 2. Eksperimen Uninformed Search (BFS vs DFS vs UCS)
Melihat perbedaan antara minimasi langkah (BFS) dengan minimasi biaya jarak kumulatif (UCS).

```python
res_bfs = breadth_first_search(romania)
res_dfs = depth_first_search(romania)
res_ucs = uniform_cost_search(romania)

print(f"BFS: {res_bfs.solution_path} | Jarak: {res_bfs.path_cost} km | Expanded: {res_bfs.nodes_expanded}")
print(f"DFS: {res_dfs.solution_path} | Jarak: {res_dfs.path_cost} km | Expanded: {res_dfs.nodes_expanded}")
print(f"UCS: {res_ucs.solution_path} | Jarak: {res_ucs.path_cost} km | Expanded: {res_ucs.nodes_expanded}")
```

## 3. Eksperimen Informed Search (Greedy Best-First vs A*)
Membandingkan pencarian serakah murni h(n) dengan pencarian optimal A* f(n) = g(n) + h(n).

```python
res_greedy = greedy_best_first_search(romania)
res_astar = astar_search(romania)

print(f"Greedy: {res_greedy.solution_path} | Jarak: {res_greedy.path_cost} km | Expanded: {res_greedy.nodes_expanded}")
print(f"A*    : {res_astar.solution_path} | Jarak: {res_astar.path_cost} km | Expanded: {res_astar.nodes_expanded}")
```

## 4. Navigasi Grid 2D dengan Rintangan
Menguji A* dengan Manhattan Heuristic pada grid 4x4 berpenghalang.

```python
grid = GridNavigationProblem((0, 0), (3, 3), rows=4, cols=4, obstacles={(1, 1), (1, 2), (2, 1)})
res_grid_astar = astar_search(grid)
print("Solusi A* Grid:", res_grid_astar.solution_path)
print("Total Biaya Langkah:", res_grid_astar.path_cost)
print("Node Diekspansi:", res_grid_astar.nodes_expanded)
```
