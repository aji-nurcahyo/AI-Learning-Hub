## 1. Pembuktian Aljabar: Normalisasi L2 Menyamakan Dot Product & Cosine Similarity
$$\hat{\mathbf{u}} = \frac{\mathbf{u}}{\|\mathbf{u}\|_2}, \quad \hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|_2} \implies \hat{\mathbf{u}} \cdot \hat{\mathbf{v}} = \cos(\theta)$$

```python
u = np.array([3.0, 4.0, 0.0])
v = np.array([1.0, 2.0, 2.0])

# Cosine Similarity langsung
cos_direct = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Normalisasi L2 ke vektor satuan
u_norm = u / np.linalg.norm(u)
v_norm = v / np.linalg.norm(v)
dot_norm = np.dot(u_norm, v_norm)

print(f'Cosine Similarity Langsung    : {cos_direct:.6f}')
print(f'Dot Product Vektor Ternormalisasi : {dot_norm:.6f}')
print(f'Selisih Mutlak               : {abs(cos_direct - dot_norm):.8e} (Terbukti Identik 100%)')
```

## 3. Metrik Evaluasi Retrieval: MRR dan Recall@k
Menghitung performa ranking kueri secara objektif.

```python
relevant_docs = {'DOC-11', 'DOC-05'}
retrieved_docs = ['DOC-11', 'DOC-12', 'DOC-05']

hits_at_3 = len(set(retrieved_docs[:3]).intersection(relevant_docs))
p_at_3 = hits_at_3 / 3.0
r_at_3 = hits_at_3 / len(relevant_docs)
rr = 1.0 / (retrieved_docs.index('DOC-11') + 1)

print(f'Precision@3 : {p_at_3*100:.2f}%')
print(f'Recall@3    : {r_at_3*100:.2f}%')
print(f'Reciprocal Rank: {rr:.4f} (Dokumen relevan pertama ada di peringkat 1)')
```
